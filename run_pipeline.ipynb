```python
from alert_grouping.pipeline import AlertGroupingPipeline
# Custom configuration for high-volume environments
# We go with default values
pipeline = AlertGroupingPipeline(
    max_time_gap=5,
    cohesion_threshold=0.3,
    entropy_threshold=1.35,
    ip_weight=0.3,
    consider_subnets=True,
    time_label =None
)
```

```python
import os
import time
import pandas as pd


def process_alert_files(
    input_dir="/home/arch/Downloads/AIT-ADS/alerts_csv",
    output_dir="/home/arch/Downloads/Alert-grouping-pipeline/output"
):
    os.makedirs(output_dir, exist_ok=True)

    results = []

    csv_files = sorted(
        f for f in os.listdir(input_dir)
        if f.endswith(".csv")
    )

    for filename in csv_files:
        input_path = os.path.join(input_dir, filename)

        # Skip previously generated output files if they are in the same folder
        if "_grouped" in filename or "_group_validity" in filename:
            continue

        print(f"\nProcessing: {filename}")

        # File size in MB
        file_size_mb = os.path.getsize(input_path) / (1024 ** 2)

        # Load data
        alerts_df = pd.read_csv(input_path)
        num_alerts = len(alerts_df)

        # Initialize pipeline
        pipeline = AlertGroupingPipeline()

        # Measure complete processing time
        start_time = time.perf_counter()

        grouped_df, validity_results = pipeline.analyze(alerts_df)

        processing_time = time.perf_counter() - start_time

        # Number of resulting groups
        num_groups = grouped_df["timely_alert_group"].nunique()

        # Output filenames
        base_name = os.path.splitext(filename)[0]

        grouped_output = os.path.join(
            output_dir,
            f"{base_name}_grouped.csv"
        )

        validity_output = os.path.join(
            output_dir,
            f"{base_name}_group_validity.csv"
        )

        # Write output datasets
        grouped_df.to_csv(grouped_output, index=False)
        validity_results.to_csv(validity_output, index=False)

        # Store summary
        results.append({
            "file": filename,
            "file_size_mb": file_size_mb,
            "num_alerts": num_alerts,
            "num_groups": num_groups,
            "processing_time_seconds": processing_time
        })

        print(
            f"  Alerts: {num_alerts:,} | "
            f"Groups: {num_groups:,} | "
            f"Size: {file_size_mb:.2f} MB | "
            f"Time: {processing_time:.2f} s"
        )

    # Summary table
    results_df = pd.DataFrame(results)

    return results_df
```

```python
results_df = process_alert_files()
```
