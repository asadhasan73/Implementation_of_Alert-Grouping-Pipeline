```python
from pathlib import Path
import pandas as pd


def analyze_alerts_csv(folder="alerts_csv"):
    folder = Path(folder)

    for file_path in sorted(folder.glob("*.csv")):
        df = pd.read_csv(file_path)

        # -------------------------------------------------
        # Determine IP to use for grouping/statistics
        #
        # Priority:
        #   1. src_ip
        #   2. agent_ip
        #   3. null
        # -------------------------------------------------

        src_ip = df["src_ip"].replace("", pd.NA)
        agent_ip = df["agent_ip"].replace("", pd.NA)

        df["grouping_ip"] = src_ip.fillna(agent_ip)

        # -------------------------------------------------
        # Statistics
        # -------------------------------------------------

        print(f"\n{file_path.name}:")
        print(f"Number of total alerts: {len(df)}")
        print(f"Number of unique hosts: {df['host'].nunique()}")
        print(
            f"Number of unique IP addresses: "
            f"{df['grouping_ip'].nunique()}"
        )
        print(f"Number of total unique names: {df['name'].nunique()}")
        print(
            f"Number of unique descriptions: "
            f"{df['short'].nunique()}"
        )
```

```python
analyze_alerts_csv()
```
